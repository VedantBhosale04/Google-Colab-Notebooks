### IRIS Dataset Description

The Iris dataset is a classic and very easy-to-understand dataset for classification. It contains measurements of sepal length, sepal width, petal length, and petal width for three species of iris flowers.

```python
# Load IRIS dataset
iris = load_iris()
iris_df = pd.DataFrame(data=np.c_[iris['data'], iris['target']],
                     columns=iris['feature_names'] + ['species'])

# Map species target to names
iris_df['species'] = iris_df['species'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})

print("IRIS Dataset Info:")
iris_df.info()
print("\nNumber of tuples (rows):", iris_df.shape[0])
print("Number of features (columns):", iris_df.shape[1])
print("\nMissing values per feature:")
display(iris_df.isnull().sum().to_frame(name='Missing Values'))

print("\nNumerical features descriptive statistics:")
display(iris_df.describe())
```

**IRIS Dataset Characteristics:**

*   **Number of features:** 5 (4 numerical measurement features + 1 categorical target feature 'species').
*   **Feature Categories:**
    *   `sepal length (cm)`: Numerical (continuous, ratio scale)
    *   `sepal width (cm)`: Numerical (continuous, ratio scale)
    *   `petal length (cm)`: Numerical (continuous, ratio scale)
    *   `petal width (cm)`: Numerical (continuous, ratio scale)
    *   `species`: Categorical (nominal, 3 unique categories: 'setosa', 'versicolor', 'virginica')
*   **Number of tuples:** 150
*   **Missing values:** None.

### IRIS Dataset: Non-Graphical EDA

For the IRIS dataset, non-graphical EDA techniques performed include:

1.  **`df.info()`**: To check data types, non-null counts, and memory usage.
2.  **`df.isnull().sum()`**: To identify and count missing values in each feature.
3.  **`df.shape`**: To determine the number of tuples (rows) and features (columns).
4.  **`df.describe()`**: To get descriptive statistics for numerical features (count, mean, std, min, 25%, 50%, 75%, max).
5.  **`df['species'].value_counts()`**: To get the frequency distribution of the categorical 'species' feature.
6.  **`df.corr()`**: To calculate the correlation matrix between numerical features.
7.  **`df.skew()` and `df.kurt()`**: To assess the skewness and kurtosis of numerical feature distributions.
8.  **`df.groupby('species').mean()` / `.median()` / `.std()`**: To compare statistics of numerical features across different species.

### Air Quality Dataset: Multivariate Analysis Graph

To perform multivariate analysis on the Air Quality dataset, several graphs can be used, especially to understand relationships between multiple pollution parameters, temperature, humidity, and location over time. A very effective graph for **multivariate analysis** is a:

*   **Pair Plot (or Scatter Plot Matrix):** This graph displays pairwise relationships between all numerical features. For each pair of features, it creates a scatter plot, and for each single feature, it shows a histogram or KDE plot. This allows for quick visualization of correlations, distributions, and potential clusters across multiple variables simultaneously. Libraries like `seaborn.pairplot` are excellent for this.

Other useful graphs for multivariate analysis in this context could include:
*   **Correlation Heatmap:** To visualize the correlation coefficients between all pairs of numerical variables.
*   **Parallel Coordinates Plot:** Effective for comparing multiple dimensions across different categories or time points.
*   **3D Scatter Plot:** If you want to visualize the relationship between exactly three numerical variables.

### Wine Quality Dataset: Non-Graphical EDA

**Yes, you can absolutely perform non-graphical EDA on the Wine Quality dataset.**

Even though the dataset primarily consists of numerical features, non-graphical methods are crucial for understanding its structure and statistical properties. Here's why and what can be done:

1.  **Descriptive Statistics (`df.describe()`):** This is foundational. It provides central tendency (mean, median), dispersion (standard deviation, min/max, quartiles), and count for all numerical features. This immediately gives insights into the range and spread of each physicochemical property and the quality ratings.
2.  **`df.info()`:** To check data types, ensure all columns are recognized as numerical, and confirm the absence of null values.
3.  **`df.isnull().sum()`:** To verify the presence or absence of missing values.
4.  **Correlation Matrix (`df.corr()`):** Since all features are numerical, calculating the pairwise correlation coefficients between all features is a powerful non-graphical EDA technique. It reveals linear relationships and their strength (e.g., how 'alcohol' content relates to 'quality', or 'fixed acidity' to 'pH').
5.  **Value Counts for 'quality' (`df['quality'].value_counts()`):** While 'quality' is numerical, it's often treated as an ordinal categorical variable. `value_counts()` will show the distribution of different quality ratings, indicating which ratings are most common.
6.  **Skewness and Kurtosis (`df.skew()`, `df.kurt()`):** These measures provide insights into the shape of the distribution of each numerical feature without plotting. Skewness indicates asymmetry, and kurtosis indicates the 'tailedness' of the distribution.
7.  **`df.groupby('quality').mean()` / `.median()` / `.std()`:** Grouping the data by 'quality' and calculating statistics for other features can show how the physicochemical properties change with different quality ratings. For example, you can see the average alcohol content or acidity for wines rated '8' versus wines rated '3'.
