### 1. When do we go for Chi-square test?

The Chi-square test is a non-parametric statistical hypothesis test primarily used when you have **categorical data** (nominal or ordinal). It helps determine if there is a statistically significant association between two categorical variables, or if an observed distribution of categories differs significantly from an expected distribution.

**Key Scenarios for using Chi-square test:**

*   **Goodness of Fit Test:** To determine if a sample distribution matches a hypothesized population distribution. For example, checking if the observed frequencies of different colors of M&Ms in a bag match the manufacturer's stated proportions.
*   **Test of Independence:** To assess if two categorical variables are independent of each other. For example, testing if there's an association between gender and preferred type of movie genre, or between treatment type and patient outcome.

### 2. Do we use Chi-square test for independence? Explain How?

Yes, the Chi-square test is very commonly used for **testing independence** between two categorical variables. This is often referred to as the Chi-square Test of Independence.

**How it works:**

1.  **Formulate Hypotheses:**
    *   **Null Hypothesis (H0):** The two categorical variables are independent (i.e., there is no association between them).
    *   **Alternative Hypothesis (H1):** The two categorical variables are not independent (i.e., there is an association between them).

2.  **Create a Contingency Table:** The first step is to organize your observed data into a contingency table (also known as a cross-tabulation table). This table displays the frequency distribution of the variables in a matrix format, where rows represent categories of one variable and columns represent categories of the other.

    *Example Contingency Table (Gender vs. Preferred Movie Genre):*

| | Action | Comedy | Drama | Total |
|---|---|---|---|---|
| **Male** | 50 | 30 | 20 | 100 |
| **Female** | 20 | 40 | 40 | 100 |
| **Total** | 70 | 70 | 60 | 200 |

3.  **Calculate Expected Frequencies:** Under the assumption that the null hypothesis (independence) is true, we calculate the expected frequency for each cell in the contingency table. The expected frequency for a cell is calculated as:
    
    `Expected Frequency = (Row Total × Column Total) / Grand Total`

    For example, the expected frequency for 'Male' and 'Action' would be (100 * 70) / 200 = 35.

4.  **Calculate the Chi-square Statistic:** The Chi-square statistic (χ²) measures the discrepancy between the observed frequencies (O) in your data and the expected frequencies (E) if the variables were truly independent. The formula is:

    `χ² = Σ [(O - E)² / E]`

    This sum is calculated across all cells in the contingency table. A larger χ² value indicates a greater difference between observed and expected frequencies, suggesting evidence against independence.

5.  **Determine Degrees of Freedom (df):** For a contingency table, `df = (number of rows - 1) × (number of columns - 1)`.

6.  **Make a Decision:** Compare the calculated χ² value with a critical value from the Chi-square distribution table (or use the p-value) at a chosen significance level (α, e.g., 0.05). If the calculated χ² exceeds the critical value (or if the p-value is less than α), you reject the null hypothesis, concluding that there is a statistically significant association between the two variables.

### 3. Explain role of Critical value and p-value in Chi-square test.

Both the critical value and the p-value are crucial for making a decision in a hypothesis test, including the Chi-square test. They serve as different ways to assess the strength of evidence against the null hypothesis.

#### Role of Critical Value:

*   **Definition:** The critical value is a threshold value obtained from the Chi-square distribution table (or software) based on the chosen significance level (alpha, α) and the degrees of freedom (df) of the test. It defines the boundary of the rejection region.
*   **Decision Rule:**
    *   If the calculated Chi-square statistic (χ²) is **greater than or equal to** the critical value, you **reject the null hypothesis (H0)**. This means your observed data is sufficiently different from what you would expect if H0 were true, indicating a statistically significant association.
    *   If the calculated Chi-square statistic (χ²) is **less than** the critical value, you **fail to reject the null hypothesis (H0)**. This means your observed data is not sufficiently different from what you would expect under H0, suggesting no statistically significant association.
*   **Intuition:** The critical value essentially marks the point beyond which the observed deviation from the expected values (under H0) is considered too extreme to have occurred by chance alone at the given significance level.

#### Role of p-value:

*   **Definition:** The p-value (probability value) is the probability of observing a test statistic as extreme as, or more extreme than, the one calculated from your sample data, assuming that the null hypothesis (H0) is true. In simpler terms, it's the probability of getting your results (or more extreme results) if there truly is no effect or no difference in the population.
*   **Decision Rule:**
    *   If the p-value is **less than or equal to** the chosen significance level (α), you **reject the null hypothesis (H0)**. This implies that the observed association is unlikely to have occurred by random chance alone.
    *   If the p-value is **greater than** the chosen significance level (α), you **fail to reject the null hypothesis (H0)**. This suggests that the observed association could reasonably have occurred by random chance.
*   **Intuition:** A small p-value means that the observed data is very unlikely if the null hypothesis were true, thus providing strong evidence against H0. A large p-value means the observed data is quite likely if H0 were true, thus providing little evidence against H0.

#### Relationship and Preference:

Both methods lead to the same conclusion for a given α. Modern statistical software typically provides the p-value, which is often preferred because it gives a more precise measure of the strength of evidence against H0 than simply comparing to a critical value.

### 4. Can we use Chi-square test to show existence of association between 2 features in a dataset? How will you interpret test result to indicate association presence.

Yes, the Chi-square test is a primary tool to show the **existence of an association (or dependence)** between two categorical features in a dataset. While it doesn't measure

**correlation**

 in the same way Pearson's correlation coefficient measures linear relationships for continuous variables, it effectively tells you if the categories of one variable are related to the categories of another.

#### Interpreting Test Results to Indicate Association Presence:

When performing a Chi-square Test of Independence, you start with the following hypotheses:

*   **Null Hypothesis (H0):** The two categorical features are independent (i.e., there is no association).
*   **Alternative Hypothesis (H1):** The two categorical features are not independent (i.e., there is an association).

Your interpretation depends on the outcome of the statistical test:

1.  **If you Reject the Null Hypothesis (H0):**
    *   This occurs when your calculated Chi-square statistic is greater than the critical value, or more commonly, when your **p-value is less than or equal to your chosen significance level (α)** (e.g., α = 0.05).
    *   **Interpretation:** You conclude that there is **statistically significant evidence of an association** between the two categorical features. This means that the observed frequencies in your contingency table are significantly different from what you would expect if the variables were truly independent. In practical terms, knowing the category of one variable gives you information about the likely category of the other variable.
    *   **What it doesn't tell you:** A significant result doesn't tell you the *strength* or *direction* of the association directly, nor does it imply causation. It only tells you that an association exists.
    *   **Further Analysis (Post-hoc):** To understand *where* the association lies, you might perform post-hoc analyses, such as examining standardized residuals for each cell in the contingency table. Large positive residuals indicate that a particular combination of categories occurs more often than expected by chance, while large negative residuals indicate it occurs less often than expected.

2.  **If you Fail to Reject the Null Hypothesis (H0):**
    *   This occurs when your calculated Chi-square statistic is less than the critical value, or when your **p-value is greater than your chosen significance level (α)**.
    *   **Interpretation:** You conclude that there is **no statistically significant evidence of an association** between the two categorical features. This suggests that any observed differences from the expected frequencies (under independence) could reasonably be attributed to random chance. In practical terms, knowing the category of one variable does not provide useful information about the likely category of the other.
    *   **Important Note:** Failing to reject H0 does not mean you have *proven* independence. It simply means your data does not provide sufficient evidence to conclude that an association exists at your chosen significance level. It's possible that a real, but subtle, association exists that your study design or sample size was not powerful enough to detect.

### 5. What is degrees of freedom and how to calculate it while performing Chi-square test.

#### What are Degrees of Freedom (df)?

In statistics, **degrees of freedom (df)** refers to the number of independent values or pieces of information that went into calculating a statistic and that are 'free to vary' in a dataset. It's often conceptualized as the number of values in a final calculation that can vary once certain constraints have been placed on the data. For instance, if you have a set of numbers that must sum to a specific total, once you know all but one of the numbers, the last number is fixed and no longer 'free to vary'.

In hypothesis testing, degrees of freedom are crucial because they determine the shape of the sampling distribution (like the Chi-square distribution). Different degrees of freedom lead to different critical values and p-values.

#### How to calculate Degrees of Freedom for a Chi-square Test of Independence:

For a Chi-square Test of Independence performed on a contingency table (a table with `r` rows and `c` columns), the degrees of freedom are calculated using the following formula:

`df = (number of rows - 1) × (number of columns - 1)`

Or, more simply:

`df = (r - 1) × (c - 1)`

**Explanation:**

Imagine a contingency table. Once you know the row totals and column totals, you don't need to fill in every cell independently. If you fill in all but one cell in each row and all but one cell in each column, the values for the remaining cells are determined by the row and column totals. The `(r-1)` represents the number of cells whose values can vary freely within each row given the row total, and `(c-1)` represents the number of cells whose values can vary freely within each column given the column total. Multiplying these together gives the total number of cells that can vary freely.

**Example:**

Consider a contingency table analyzing the relationship between "Gender" (Male, Female) and "Vote Preference" (Candidate A, Candidate B, Candidate C).

*   **Number of rows (r):** 2 (Male, Female)
*   **Number of columns (c):** 3 (Candidate A, Candidate B, Candidate C)

Using the formula:

`df = (r - 1) × (c - 1)`
`df = (2 - 1) × (3 - 1)`
`df = (1) × (2)`
`df = 2`

So, for this example, there are 2 degrees of freedom.
