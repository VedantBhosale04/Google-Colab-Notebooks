## GRAPHICAL ANALYSIS 1 (Univariate)
Objective: To analyze distribution of Age  
Method: Histogram  

Observation: Most passengers are between 20–40 years  
Conclusion: Majority are young adults  



```python
import matplotlib.pyplot as plt
import seaborn as sns

# Histogram
plt.hist(df['Age'])
plt.title("Age Distribution")
plt.show()
```

## GRAPHICAL ANALYSIS 2 (Multivariate)
Objective: Survival based on Gender  
Method: Countplot  

Observation: Females survived more than males  
Conclusion: Gender plays an important role  



```python
# Countplot
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title("Survival based on Gender")
plt.show()
```
