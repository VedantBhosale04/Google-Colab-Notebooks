## 1. What is a 2-tailed and 1-tailed T-Test?

A **T-test** is a statistical hypothesis test that compares the means of two groups. It is used to determine if there is a significant difference between the means of these groups.

The terms "1-tailed" and "2-tailed" refer to the directionality of the hypothesis being tested.

### 2-tailed T-Test

*   **Purpose:** A 2-tailed (or two-sided) test is used when you are interested in detecting a difference between two groups in *either direction*. You are hypothesizing that the means are different, but you don't specify whether one mean is greater or smaller than the other.
*   **Null Hypothesis (H0):** There is no significant difference between the two group means (e.g., Mean A = Mean B).
*   **Alternative Hypothesis (Ha):** There is a significant difference between the two group means (e.g., Mean A ≠ Mean B).
*   **Critical Region:** The critical region for rejecting the null hypothesis is split into two tails of the t-distribution (one on each side). If your calculated t-statistic falls into either of these tails, you reject H0.
*   **When to use:** Use a 2-tailed test when you want to know if there's *any* difference, without making an a priori assumption about the direction of that difference. For example, comparing the effectiveness of a new drug versus a placebo, where the drug could be better *or* worse.

### 1-tailed T-Test

*   **Purpose:** A 1-tailed (or one-sided) test is used when you have a specific directional hypothesis about the difference between two groups. You are hypothesizing that one mean is specifically greater than the other, or specifically less than the other.
*   **Null Hypothesis (H0):** Depending on the direction, e.g., Mean A ≤ Mean B (for testing if Mean A > Mean B) or Mean A ≥ Mean B (for testing if Mean A < Mean B).
*   **Alternative Hypothesis (Ha):** Depending on the direction, e.g., Mean A > Mean B (right-tailed) or Mean A < Mean B (left-tailed).
*   **Critical Region:** The critical region is entirely located in one tail of the t-distribution, corresponding to the direction of your alternative hypothesis. If your calculated t-statistic falls into this single tail, you reject H0.
*   **When to use:** Use a 1-tailed test when you have a strong theoretical or empirical reason to expect a difference in a particular direction. For example, if you are testing if a new fertilizer *increases* crop yield, you would use a right-tailed test.

### Key Difference

The primary difference lies in the alternative hypothesis and how the p-value is interpreted. A 1-tailed test has more statistical power to detect a difference in the specified direction, but it will miss a difference in the opposite direction. A 2-tailed test is more conservative but can detect differences in either direction.

## 2. Explain what are Type I and Type II errors with an example.

In hypothesis testing, we make a decision about the null hypothesis (H0) based on sample data. Since we are dealing with probabilities and samples, there's always a chance of making an incorrect decision. These incorrect decisions are categorized into two types of errors:

### Type I Error (False Positive)

*   **Definition:** A Type I error occurs when you **reject a true null hypothesis**.
*   **Analogy:** It's like a false alarm. You conclude there's an effect or difference when, in reality, there isn't one.
*   **Symbol:** The probability of committing a Type I error is denoted by **α (alpha)**, which is also the significance level you set for your test (e.g., 0.05 or 5%).
*   **Consequence:** Often involves unnecessary changes, costs, or further research based on a non-existent effect.

**Example:**
Imagine a pharmaceutical company is testing a new drug for a rare disease. The null hypothesis (H0) is that "The new drug has no effect on the disease." The alternative hypothesis (Ha) is "The new drug has an effect on the disease."

A Type I error would occur if the company concludes that the drug is effective (rejects H0), but in reality, the drug has no effect. This could lead to the drug being approved and prescribed to patients, potentially causing side effects without any therapeutic benefit.

### Type II Error (False Negative)

*   **Definition:** A Type II error occurs when you **fail to reject a false null hypothesis**.
*   **Analogy:** It's like missing a real alarm. You conclude there's no effect or difference when, in reality, there *is* one.
*   **Symbol:** The probability of committing a Type II error is denoted by **β (beta)**.
*   **Consequence:** Can lead to missed opportunities, failure to implement beneficial changes, or continued suffering if an effective treatment is overlooked.

**Example:**
Using the same drug example:

A Type II error would occur if the company concludes that the drug has no effect (fails to reject H0), but in reality, the drug *is* effective. This could lead to a potentially life-saving or beneficial drug being discarded, preventing patients from receiving effective treatment.

### Summary Table

|                   | **Actual State of H0** |
| :---------------- | :--------------------- | :---------------------- |
| **Decision**      | **H0 is True**         | **H0 is False**         |
| **Reject H0**     | Type I Error (α)       | Correct Decision (Power) |
| **Fail to Reject H0** | Correct Decision       | Type II Error (β)       |

## 3. What is power of statistical test?

The **power of a statistical test** is the probability of correctly rejecting a false null hypothesis. In simpler terms, it's the probability that your test will detect an effect or difference when an effect or difference actually exists in the population.

*   **Relationship to Type II Error:** Power is directly related to the Type II error (β). Specifically, **Power = 1 - β**. If a test has high power, it means it has a low probability of committing a Type II error.

*   **Goal of High Power:** Researchers generally aim for tests with high power (typically 0.80 or 80%) to increase the chances of detecting a true effect. A power of 0.80 means there's an 80% chance of detecting a real effect if it exists, and a 20% chance of missing it (Type II error).

*   **Factors Influencing Power:**
    1.  **Significance Level (α):** Increasing α (e.g., from 0.05 to 0.10) increases power, but also increases the risk of a Type I error.
    2.  **Effect Size:** This is the magnitude of the difference or relationship you are trying to detect. A larger effect size is easier to detect, thus increasing power. Smaller effects require more power to be detected.
    3.  **Sample Size (N):** Increasing the sample size generally increases the power of a test because larger samples provide more information and reduce sampling variability, making it easier to detect true effects.
    4.  **Variability (Standard Deviation):** Lower variability within the data (smaller standard deviation) leads to higher power, as it makes the effect clearer against the background noise.

## 4. Explain input and output characteristics for `stats.ttest_rel` function in your own words.

The `scipy.stats.ttest_rel` function in Python is used to perform a **dependent (or paired) samples t-test**. This test is appropriate when you have two sets of measurements that are related or paired in some way, such as measurements taken from the same subjects before and after an intervention, or measurements from matched pairs.

Its main purpose is to determine if the mean difference between these paired observations is significantly different from zero.

Let's look at its typical use and characteristics:

```python
import scipy.stats as stats

# Example: Blood pressure before and after a treatment for the same patients
before_treatment = [120, 125, 130, 122, 128, 135, 129, 121, 127, 133]
after_treatment = [115, 120, 125, 120, 126, 130, 124, 118, 125, 130]

# Perform the paired t-test
t_statistic, p_value = stats.ttest_rel(before_treatment, after_treatment)

print(f"T-statistic: {t_statistic:.3f}")
print(f"P-value: {p_value:.3f}")
```

### Input Characteristics (`stats.ttest_rel(a, b, axis=0, nan_policy='propagate')`)

1.  **`a` (array_like):**
    *   **What it is:** This is the first set of measurements or observations. Think of it as the 'before' data or the first measurement in a pair.
    *   **Format:** It should be an array-like object (like a list or a NumPy array) containing numerical data.
    *   **Requirement:** It *must* have the same shape and size as `b`, because the test compares each element of `a` directly with the corresponding element of `b`.

2.  **`b` (array_like):**
    *   **What it is:** This is the second set of measurements or observations. Think of it as the 'after' data or the second measurement in a pair.
    *   **Format:** Similar to `a`, it should be an array-like object containing numerical data.
    *   **Requirement:** It *must* have the same shape and size as `a` for the paired comparison.

3.  **`axis` (int or None, optional):**
    *   **What it is:** This argument specifies which axis of the input arrays `a` and `b` to operate along, if they are multi-dimensional.
    *   **Default:** `0` (meaning it operates column-wise for 2D arrays).
    *   **Significance:** If you have multiple columns of paired data (e.g., several different measurements for each patient), `axis` tells the function how to pair them up. If `None`, the arrays are flattened before computing the test.

4.  **`nan_policy` ({'propagate', 'raise', 'omit'}, optional):**
    *   **What it is:** This handles how the function should behave if there are `NaN` (Not a Number) values in your input data.
    *   **`'propagate'` (default):** If a `NaN` is present, `NaN` will be returned for the statistic and p-value. This means if even one `NaN` exists, the result for that pair (or the entire calculation if a single `NaN` affects it) becomes `NaN`.
    *   **`'raise'`:** If a `NaN` is encountered, a `ValueError` will be raised, stopping the execution.
    *   **`'omit'`:** Pairs containing `NaN` values will be removed from the calculation. This is useful if you want to perform the test on all available complete pairs, but be aware it reduces your sample size.

### Output Characteristics (`statistic, pvalue`)

1.  **`statistic` (float or array):**
    *   **What it is:** This is the calculated t-statistic. It's a measure of how far the observed sample mean difference is from the hypothesized mean difference of zero, relative to the variability of the differences.
    *   **Interpretation:** A larger absolute value of the t-statistic suggests a larger difference between the means, relative to the standard error of the difference.
    *   **Range:** It can be any real number (positive or negative).

2.  **`pvalue` (float or array):**
    *   **What it is:** This is the two-tailed p-value associated with the calculated t-statistic.
    *   **Interpretation:** The p-value tells you the probability of observing a t-statistic as extreme as, or more extreme than, the one calculated, *assuming the null hypothesis is true* (i.e., assuming there is no true mean difference).
    *   **Range:** The p-value always ranges from **0 to 1**.
        *   **Small p-value (typically < α, e.g., < 0.05):** Suggests that the observed difference is unlikely to have occurred by random chance alone if the null hypothesis were true. This leads to rejecting the null hypothesis and concluding there is a statistically significant difference.
        *   **Large p-value (typically ≥ α):** Suggests that the observed difference could reasonably have occurred by random chance. This leads to failing to reject the null hypothesis, meaning there isn't enough evidence to conclude a statistically significant difference.

In essence, `ttest_rel` takes your paired data, calculates how much the two sets differ on average, and then tells you how likely that observed average difference is if there was actually no difference in the population.

````python
import scipy.stats as stats
import numpy as np

# Example data: Test scores of 10 students before and after a tutoring program
before_scores = np.array([75, 80, 65, 90, 70, 85, 78, 92, 68, 73])
after_scores = np.array([80, 85, 70, 95, 72, 88, 80, 90, 75, 78])

print("Before Scores:", before_scores)
print("After Scores:", after_scores)

# Perform the paired samples t-test
t_statistic, p_value = stats.ttest_rel(before_scores, after_scores)

print(f"\nCalculated T-statistic: {t_statistic:.4f}")
print(f"Calculated P-value: {p_value:.4f}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print(f"With p-value ({p_value:.4f}) < alpha ({alpha}), we reject the null hypothesis.")
    print("Conclusion: There is a statistically significant difference in scores before and after the tutoring program.")
else:
    print(f"With p-value ({p_value:.4f}) >= alpha ({alpha}), we fail to reject the null hypothesis.")
    print("Conclusion: There is no statistically significant difference in scores before and after the tutoring program.")

# Example with NaN policy 'omit'
before_with_nan = np.array([75, 80, np.nan, 90, 70])
after_with_nan = np.array([80, 85, 70, 95, 72]) # One of these will be omitted along with the NaN in 'before'

print("\n--- Demonstrating nan_policy='omit' ---")
print("Before Scores (with NaN):", before_with_nan)
print("After Scores (paired with NaN data):", after_with_nan)

t_statistic_omit, p_value_omit = stats.ttest_rel(before_with_nan, after_with_nan, nan_policy='omit')

print(f"Calculated T-statistic (omit NaN): {t_statistic_omit:.4f}")
print(f"Calculated P-value (omit NaN): {p_value_omit:.4f}")
````

## 5. What is function in stats module for identifying power of a statistical test. Explain output range of it and its significance.

While `scipy.stats` provides a wide range of statistical functions, it does **not have a dedicated, direct function within its core for identifying the power of a statistical test.** Power analysis is typically performed using the `statsmodels` library in Python, specifically its `stats.power` submodule, which offers more comprehensive tools for this purpose.

One of the most commonly used functions for power analysis, especially for t-tests, is `statsmodels.stats.power.TTestPower`.

Let's demonstrate how to use `TTestPower` and explain its output.

```python
import statsmodels.stats.power as smp
import matplotlib.pyplot as plt
import numpy as np

# 1. Initialize the power analysis object for a one-sample or two-sample t-test
power_analysis = smp.TTestPower()

# 2. Calculate power given effect size, sample size, and significance level
# Example scenario: Detecting a medium effect size (d=0.5) with 30 observations per group
# and an alpha of 0.05 for a two-sided test.

effect_size = 0.5  # Cohen's d: 0.2 small, 0.5 medium, 0.8 large
sample_size_per_group = 30
alpha = 0.05       # Significance level

power = power_analysis.solve_power(
    effect_size=effect_size,
    nobs1=sample_size_per_group,
    alpha=alpha,
    power=None, # We want to solve for power
    ratio=1.0,  # Ratio of sample sizes (nobs2/nobs1). 1.0 for equal groups
    alternative='two-sided' # 'two-sided', 'less', 'greater'
)

print(f"Calculated Power: {power:.4f}")

# 3. Determine required sample size for a desired power
desired_power = 0.80 # 80% chance of detecting a true effect

required_n = power_analysis.solve_power(
    effect_size=effect_size,
    nobs1=None, # We want to solve for nobs1 (sample size)
    alpha=alpha,
    power=desired_power,
    ratio=1.0,
    alternative='two-sided'
)

print(f"\nRequired sample size (per group) for {desired_power*100:.0f}% power: {np.ceil(required_n):.0f}")

# 4. Plotting power curves (how power changes with sample size)
sample_sizes = np.arange(10, 100, 5) # Sample sizes from 10 to 95
powers = [power_analysis.solve_power(effect_size=0.5, nobs1=n, alpha=0.05, power=None, alternative='two-sided') for n in sample_sizes]

plt.figure(figsize=(10, 6))
plt.plot(sample_sizes, powers, marker='o')
plt.title('Power vs. Sample Size (Effect Size = 0.5, Alpha = 0.05)')
plt.xlabel('Sample Size per Group (n)')
plt.ylabel('Power')
plt.grid(True)
plt.axhline(y=0.8, color='r', linestyle='--', label='Desired Power (0.8)')
plt.legend()
plt.show()
```

### Output Range and Significance of Power from `statsmodels.stats.power`

When you use functions like `solve_power` from `statsmodels.stats.power`, the output for `power` will be a single float value.

*   **Output Range:** The power value will always be between **0 and 1 (inclusive)**.
    *   `0`: Indicates no chance of detecting an effect, even if it truly exists.
    *   `1`: Indicates a 100% chance of detecting an effect, if it truly exists.

*   **Significance:**
    *   **Closer to 1 (e.g., 0.80 or 0.90):** A high power value is desirable. It means that if a true effect of the specified size exists in the population, your study has a high probability of detecting it as statistically significant. This reduces the risk of a Type II error (false negative).
    *   **Closer to 0 (e.g., 0.20 or 0.30):** A low power value indicates that your study has a low probability of detecting a true effect, even if one is present. This means you are at a high risk of committing a Type II error, potentially missing an important finding. Studies with low power are often considered underpowered and may not be worthwhile.

**In summary, the power output quantifies the sensitivity of your statistical test to detect a true effect, given your sample size, effect size, and significance level. It's a crucial metric for study design and interpretation of non-significant results.**
